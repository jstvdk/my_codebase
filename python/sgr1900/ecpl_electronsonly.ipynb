```python
import numpy as np
from astropy.io import ascii

import naima
import custom_models
```

```python
filename = 'data/SED2_HESS_10_TeV.txt'

data = ascii.read(filename, format='ipac')
```

```python

## Set initial parameters and labels
p0 = np.array((1e38, np.log10(0.5), 3.0, np.log10(30)))
labels = ["norm", "log10(ref_energy)","index", "log10(cutoff)"]

## Run sampler
sampler, pos = naima.run_sampler(
    data_table=data,
    p0=p0,
    labels=labels,
    model=custom_models.ElectronIC,
    prior=custom_models.lnprior_electrons,
    nwalkers=64,
    nburn=2000,
    nrun=8000,
    threads=5,
    prefit=False,
)

## Save run results to HDF5 file (can be read later with naima.read_run)
```

```python
naima.save_run("results/5000run_SGR_IC_run.hdf5", sampler)

## Diagnostic plots with labels for the metadata blobs
naima.save_diagnostic_plots(
    "results/5000run_SGR_IC",
    sampler,
    sed=True,
    last_step=False,
    blob_labels=[
        "Spectrum",
        "Electron energy distribution",
        "$W_e (E_e>1\, \mathrm{KeV})$",
    ],
)
naima.save_results_table("results/5000run_SGR_IC", sampler)
```
