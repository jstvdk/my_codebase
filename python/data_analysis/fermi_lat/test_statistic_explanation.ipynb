```python
import numpy as np
```

```python
def test_statistic(lambda1, lambda0):
    """
    Calculate the test statistic for a given lambda1 and lambda0.
    
    Parameters:
    lambda1 (float): The first parameter of the test statistic.
    lambda0 (float): The second parameter of the test statistic.
    
    Returns:
    float: The calculated test statistic.
    """
    return 2*(lambda1 * np.log(lambda1 / lambda0) - (lambda1 - lambda0))
```

```python
test_statistic(60, 25)
```
