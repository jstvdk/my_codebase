```python
from astropy.coordinates import SkyCoord
import astropy.units as u

# Define equatorial coordinates (J2000)
ra = "14h43m2.16s"
dec = "-62d26m56s"

# Create SkyCoord object
equatorial = SkyCoord(ra, dec, frame='icrs')

# Convert to Galactic coordinates
galactic = equatorial.galactic

# Print Galactic longitude (l) and latitude (b)
print(f"Galactic Longitude (l): {galactic.l.degree:.6f}°")
print(f"Galactic Latitude (b): {galactic.b.degree:.6f}°")
```

```python
def arcmin_to_deg(arcmin: float) -> float:
    """
    Convert arcminutes to degrees.

    Parameters
    ----------
    arcmin : float
        Angle in arcminutes.

    Returns
    -------
    float
        Angle in degrees.
    """
    return arcmin / 60.0

def deg_to_arcmin(deg: float) -> float:
    """
    Convert degrees to arcminutes.

    Parameters
    ----------
    deg : float
        Angle in degrees.

    Returns
    -------
    float
        Angle in arcminutes.
    """
    return deg * 60.0

# Example usage
print(f"8 arcmin = {arcmin_to_deg(8):.4f} degrees")
print(f"20 arcmin = {arcmin_to_deg(20):.4f} degrees")
print(f"2.5 degrees = {deg_to_arcmin(2.5):.1f} arcmin")
```
