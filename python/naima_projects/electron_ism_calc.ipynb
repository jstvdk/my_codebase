```python
import numpy as np
import astropy.units as u

def cooling_time(E: u.Quantity,
                 B: u.Quantity = 3*u.uG,
                 n_H: u.Quantity = 1*u.cm**-3,
                 photon_fields=['CMB']):
    """Electron cooling time [s] including synchrotron, IC, brems, Coulomb."""
    return 2*u.s

def diffusion_coefficient(E: u.Quantity,
                          D0: u.Quantity = 3e28*u.cm**2/u.s,
                          δ: float = 0.5,
                          E_ref: u.Quantity = 1*u.GeV):
    """Compute the diffusion coefficient D(E) = D0 * (E/E_ref)^δ."""
    return D0 * (E/E_ref)**δ

def diffusion_time(E: u.Quantity,
                   L: u.Quantity = 1*u.kpc,
                   D0: u.Quantity = 3e28*u.cm**2/u.s,
                   δ: float = 0.5,
                   E_ref: u.Quantity = 1*u.GeV):
    """Estimate the diffusion time: t_diff = L^2 / (6 * D(E))."""
    D = diffusion_coefficient(E, D0, δ, E_ref)
    return L**2 / (6 * D)

def diffusion_length(E: u.Quantity,
                     t: u.Quantity,
                     D0: u.Quantity = 3e28*u.cm**2/u.s,
                     δ: float = 0.5,
                     E_ref: u.Quantity = 1*u.GeV):
    """Compute the distance diffused in time t: r_diff = sqrt(6 * D(E) * t)."""
    D = diffusion_coefficient(E, D0, δ, E_ref)
    return np.sqrt(6 * D * t)

# Example usage:
if __name__ == '__main__':
    E = 10 * u.GeV
    ct = cooling_time(E, B=5*u.uG, n_H=0.5*u.cm**-3)
    dt = diffusion_time(E, L=0.5*u.kpc)
    dl = diffusion_length(E, ct)
    
    print(f"Cooling time @ {E}: {ct.to(u.Myr):.2f}")
    print(f"Diffusion time over 0.5 kpc: {dt.to(u.Myr):.2f}")
    print(f"Diffusion length in cooling time: {dl.to(u.kpc):.2f}")
```
