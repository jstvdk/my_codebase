```python
from astropy.coordinates import SkyCoord
import astropy.units as u

from astropy.coordinates import Angle

def convert_to_decimal(ra_hms, dec_dms):
    """
    Convert celestial coordinates from sexagesimal format to decimal degrees in the J2000 coordinate system.

    Parameters:
        ra_hms (str): Right Ascension in sexagesimal format, e.g. '12 30 00' or '12:30:00'
        dec_dms (str): Declination in sexagesimal format, e.g. '+15 20 30' or '+15:20:30'
    
    Returns:
        tuple: (ra_deg, dec_deg) where ra_deg is the right ascension in decimal degrees,
               and dec_deg is the declination in decimal degrees.
    """
    # Create a SkyCoord object in the FK5 system with the J2000.0 equinox.
    coord = SkyCoord(ra=ra_hms, dec=dec_dms, unit=(u.hourangle, u.deg),
                     frame='fk5', equinox='J2000.0')
    
    # Return the right ascension and declination in decimal degrees.
    return coord.ra.deg, coord.dec.deg

```

```python
ra_decimal, dec_decimal = convert_to_decimal("14 40 00", "60 00 00")
print("RA (deg):", ra_decimal)
print("Dec (deg):", dec_decimal)
```
