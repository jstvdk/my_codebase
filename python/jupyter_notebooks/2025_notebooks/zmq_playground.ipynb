```python
import zmq
```

```python
# Context is the container for all sockets in a single process
context = zmq.Context()

# Create a publisher socket
pub = context.socket(zmq.PUB)
pub.bind("tcp://*:5556")

# Create a subscriber socket
sub = context.socket(zmq.SUB)
sub.connect("tcp://localhost:5556")
sub.setsockopt_string(zmq.SUBSCRIBE, "weather")  # filter on "weather" topic

# Publisher side
topic = "weather"
message = "rain 10mm"
pub.send_string(f"{topic} {message}")

# Subscriber side
topic_received, body = sub.recv_string().split(' ', 1)
print(f"Received {topic_received}: {body}")
```
